```python
import pandas as pd
from pyteomics import mass
from brainpy import isotopic_variants

def get_BrIS_sequence(sequence):
    aa_list = list(sequence)
    aa_list.insert(aa_list.index('F') + 1, '(4-Br)')
    BrIS_sequence = ''.join(aa_list)
    return BrIS_sequence

def cal_BrIS_isotope_mass(sequence, nIsotopes):
    iso0_num = 79
    seq_mass = mass.calculate_mass(sequence = sequence)
    H_mass = mass.calculate_mass(composition={'H': 1})
    iso0_mass = seq_mass - H_mass + mass.calculate_mass(composition={"Br[%d]" % iso0_num: 1})
    isotopes_mass = [iso0_mass, ]
    for i in range(1, nIsotopes):
        iso_num = iso0_num + i
        iso_mass = seq_mass - H_mass + mass.calculate_mass(composition={"Br[%d]" % iso_num: 1})
        isotopes_mass.append(iso_mass)
    return isotopes_mass
    
def cal_BrIS_isotope_intensity_distribution(sequence, nIsotopes):
    composition = dict(mass.Composition(sequence = sequence))
    composition['H'] = composition['H'] - 1
    composition['Br'] = 1
    isotope_pattern = isotopic_variants(composition, npeaks = nIsotopes)
    isotopes_intensity_distribution = [isotope_pattern[i].intensity for i in range(nIsotopes)]
    return isotopes_intensity_distribution
```

```python
peptides = {'1':'TPENFPSK', '2':'LPPTTTFQQQK', '3':'ADIGFTPGSGK', '4':'DGYNNNIHFLTK',  
            '5':'EAAQDFEVLGQR', '6':'GQYIFVATSVAGEK', '7':'YTFLPGYVR',   '8':'ASYLDFIR', '9':'WELFDIPR'}
    
RTs = {'BrIS-1':12.1165, 'BrIS-2':15.5729, 'BrIS-3':16.9629, 'BrIS-4':18.5791, 
       'BrIS-5':20.3957, 'BrIS-6':22.4872, 'BrIS-7':23.9436, 'BrIS-8':25.6237, 'BrIS-9':26.5539}
nIsotopes = 3
BrIS_lib = {}
for idx, sequence in peptides.items():
    BrIS_sequence = get_BrIS_sequence(sequence)
    isotopes_mass = cal_BrIS_isotope_mass(sequence, nIsotopes)
    isotopes_intensity_distribution = cal_BrIS_isotope_intensity_distribution(sequence, nIsotopes)
    BrIS_lib[f"BrIS-{idx}"] = {'sequence':BrIS_sequence, 
                               'isotopes mass': isotopes_mass, 
                               'isotopes intensity distribution': isotopes_intensity_distribution,
                               'RT':RTs[f"BrIS-{idx}"]}
BrIS_lib = pd.DataFrame(BrIS_lib).T
BrIS_lib.index.name = 'accession'
BrIS_lib
#BrIS_lib.to_csv("c:/Users/chenxin/Desktop/BrIS_library.csv")
```
